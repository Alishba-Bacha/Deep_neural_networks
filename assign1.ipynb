$\textbf{Linear functions}$

We will be using the term linear equation to mean a weighted sum of inputs plus an offset. If there is just one input $x$, then this is a straight line:
$$
y = b + w x,
$$
where $b$ is the $y$-intercept of the linear and $w$ is the slope of the line. When there are two inputs $x_1$ and $x_2$, then this becomes:
$$
y = b + w_1 x_1 + w_2 x_2 .
$$

Any other functions are by definition non-linear.

#### $\color{blue}{\text{Define a linear function with just one input }}x$ $\hspace{15.5cm}\color{blue}{(\text{Marks }} 02)$

````python
def linfunc_1D(x,b,w):
    # TASK_1D_1 -- Formulate the 1D linear equation.
    #              Uncomment and insert your code (no more than 1 line)
    #------------------------------------------------------------------
    y = b + w*x
    # -----------------------------------------------------------------
    return y
````

#### $\color{blue}{\text{2.4 - Normalizing Rows}}$ $\hspace{16.5cm}\color{blue}{(\text{Marks }} 2)$

A common technique used in Machine Learning and Deep Learning is data normalization. This process often improves performance because gradient descent converges more quickly after normalization. In this context, normalization refers to transforming \( x \) to \( \frac{x}{\| x \|} \), where each row vector of \( x \) is divided by its norm.

For example, if we have

$$ x =
\begin{bmatrix}
    0 & 3 & 4 \\
    2 & 6 & 4 \\
\end{bmatrix} \tag{3} $$

then the norm is calculated as

$$ \| x \| = np.linalg.norm(x, \text{axis} = 1, \text{keepdims} = \text{True}) = \begin{bmatrix}
    5 \\
    \sqrt{56} \\
\end{bmatrix} \tag{4} $$

and the normalized matrix is given by

$$ x_{\text{normalized}} = \frac{x}{\| x \|} = \begin{bmatrix}
    0 & \frac{3}{5} & \frac{4}{5} \\
    \frac{2}{\sqrt{56}} & \frac{6}{\sqrt{56}} & \frac{4}{\sqrt{56}} \\
\end{bmatrix} \tag{5} $$

It's important to note that you can divide matrices of different sizes, thanks to a technique called broadcasting, which you will learn more about in part 5.

**Exercise**: Implement the function `normalizeRows()` to normalize the rows of a matrix. After applying this function to the input matrix \( x \), each row should be a vector of unit length (i.e., length 1).


````python
def normalizeRows(x):

# Hint 1: Calculate the norm of each row
  norms = np.linalg.norm(x, axis=1, keepdims=True)

# Hint 2: Normalize the rows by dividing each row by its norm
  x_normalized = x / norms

# Hint 3: Ensure the function returns the normalized matrix
  return x_normalized
````

#### $\color{blue}{\text{2.5 - Broadcasting and the Softmax Function}}$ $\hspace{16.5cm}\color{blue}{(\text{Marks }} 2)$

An important concept to grasp in numpy is "broadcasting." This feature is very useful for performing mathematical operations between arrays of different shapes. For full details on broadcasting, you can read the official [broadcasting documentation](http://docs.scipy.org/doc/numpy/user/basics.broadcasting.html).

**Exercise**: Implement a softmax function using numpy. The softmax function acts as a normalizing function when your algorithm needs to classify two or more classes. More about softmax will be covered in the second course of this specialization.

**Instructions**:
- $ \text{for } x \in \mathbb{R}^{1\times n} \text{,     } softmax(x) = softmax(\begin{bmatrix}
    x_1  &&
    x_2 &&
    ...  &&
    x_n  
\end{bmatrix}) = \begin{bmatrix}
     \frac{e^{x_1}}{\sum_{j}e^{x_j}}  &&
    \frac{e^{x_2}}{\sum_{j}e^{x_j}}  &&
    ...  &&
    \frac{e^{x_n}}{\sum_{j}e^{x_j}}
\end{bmatrix} $

- $\text{for a matrix } x \in \mathbb{R}^{m \times n} \text{,  $x_{ij}$ maps to the element in the $i^{th}$ row and $j^{th}$ column of $x$, thus we have: }$  $$softmax(x) = softmax\begin{bmatrix}
    x_{11} & x_{12} & x_{13} & \dots  & x_{1n} \\
    x_{21} & x_{22} & x_{23} & \dots  & x_{2n} \\
    \vdots & \vdots & \vdots & \ddots & \vdots \\
    x_{m1} & x_{m2} & x_{m3} & \dots  & x_{mn}
\end{bmatrix} = \begin{bmatrix}
    \frac{e^{x_{11}}}{\sum_{j}e^{x_{1j}}} & \frac{e^{x_{12}}}{\sum_{j}e^{x_{1j}}} & \frac{e^{x_{13}}}{\sum_{j}e^{x_{1j}}} & \dots  & \frac{e^{x_{1n}}}{\sum_{j}e^{x_{1j}}} \\
    \frac{e^{x_{21}}}{\sum_{j}e^{x_{2j}}} & \frac{e^{x_{22}}}{\sum_{j}e^{x_{2j}}} & \frac{e^{x_{23}}}{\sum_{j}e^{x_{2j}}} & \dots  & \frac{e^{x_{2n}}}{\sum_{j}e^{x_{2j}}} \\
    \vdots & \vdots & \vdots & \ddots & \vdots \\
    \frac{e^{x_{m1}}}{\sum_{j}e^{x_{mj}}} & \frac{e^{x_{m2}}}{\sum_{j}e^{x_{mj}}} & \frac{e^{x_{m3}}}{\sum_{j}e^{x_{mj}}} & \dots  & \frac{e^{x_{mn}}}{\sum_{j}e^{x_{mj}}}
\end{bmatrix} = \begin{pmatrix}
    softmax\text{(first row of x)}  \\
    softmax\text{(second row of x)} \\
    ...  \\
    softmax\text{(last row of x)} \\
\end{pmatrix} $$

````python
def softmax(x):
# Hint 1: Calculate the exponential of each element in x
  exp_x = np.exp(x)

# Hint 2: Compute the sum of exponentials for each row
  sum_exp_x = np.sum(exp_x, axis=1, keepdims=True)

# Hint 3: Normalize the exponentials by the sum to get softmax
  softmax_output = exp_x / sum_exp_x

# Hint 4: Ensure the function returns the softmax values
  return softmax_output
````

### 3. Vectorization in Deep Learning

Vectorization is a crucial concept in deep learning and numerical computing that involves converting operations from iterative loops into matrix and vector operations. This approach significantly enhances computational efficiency, especially when dealing with large datasets.

In traditional programming, you might use loops to perform calculations on each element of an array. However, this can be slow and computationally expensive. Vectorization allows you to apply operations on entire arrays at once, leveraging optimized libraries like NumPy that are built to handle such operations efficiently.

For example, instead of computing the sum of two arrays element-wise in a loop, you can simply use the `+` operator between the two arrays, and NumPy will handle the underlying iteration in a highly optimized manner. This leads to cleaner code and faster execution.

**Exercise**: Implement the NumPy vectorized version of the L1 loss function. The `abs(x)` function (which computes the absolute value of \( x \)) will be useful for this task.

**Reminder**:
- Loss functions are essential for assessing the performance of your model. A larger loss indicates that your predictions \( \hat{y} \) are further from the true values \( y \).
- In deep learning, optimization algorithms like Gradient Descent are used to train the model by minimizing this cost.

The L1 loss is mathematically defined as:

$$\begin{align*} & L_1(\hat{y}, y) = \sum_{i=0}^m|y^{(i)} - \hat{y}^{(i)}| \end{align*}\tag{6}$$

where:
- m is the number of samples,
- 𝑦(𝑖) is the true value for the

𝑡

  sample, and
- y^(i)
  is the predicted value for the
𝑖𝑡ℎ

**Implementation Steps**:
1. Take two input arrays, \( \hat{y} \) (predicted values) and \( y \) (true values).
2. Calculate the absolute differences between these arrays.
3. Sum all the absolute differences to compute the L1 loss.

````python
# Hint: Use numpy's absolute function and sum function
def L1_loss(y_true, y_pred):
    return np.sum(np.abs(y_true - y_pred))

# Example Usage:
y_true = np.array([1.0, 2.0, 3.0])
y_pred = np.array([1.5, 2.5, 3.5])
print(L1_loss(y_true, y_pred))  # Output should be the L1 loss value
````
