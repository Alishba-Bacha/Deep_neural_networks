## Section 2: Regularization Techniques <span style='color:blue'><strong>(20 Marks)</strong></span>

### 2.1 Problem Statement

- Overfitting occurs when your model performs well on training data but poorly on new data. Regularization techniques like L2 regularization and dropout can help prevent overfitting.
### 2.2 L2 Regularization  
**Theoretical Explanation**  
- L2 regularization adds a penalty term to the loss function, discouraging large weights in the model. This results in smaller, more generalizable weights.


#### Problem 4: Compute Cost with Regularization <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Modify the cost function to include the L2 regularization term. This will penalize large weights and help generalize the model.  
**Exercise:** Implement `compute_cost_with_regularization()`.


```python
# You need to add the regularization term to the cost function.
# Hint: Use np.sum() to compute the sum of squared weights (L2 regularization term).
# Remember to multiply by the regularization parameter lambda and divide by (2*m).

def compute_cost_with_regularization(y_true, y_pred, parameters, lambda_reg):
  m = y_true.shape[0]
  cost = (-1/m) * np.sum(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
  regularization_term = 0
  for key in parameters.keys():
    if 'W' in key:
      regularization_term += np.sum(np.square(parameters[key]))

  regularization_cost = (lambda_reg / (2 * m)) * regularization_term
  total_cost = cost + regularization_cost

  return total_cost
```

#### Problem 5: Backward Propagation with Regularization <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** In the backward propagation step, adjust the gradients to account for the L2 regularization term.  
**Exercise:** Implement `backward_propagation_with_regularization()`.

```python
# In the gradients for the weight parameters, add the L2 regularization term.
# Hint: Don't forget to include (lambda/m) * W in the gradient calculation for dW.

def backward_propagation_with_regularization(X, y_true, y_pred, parameters, caches, lambda_reg):
    m = y_true.shape[0]
    grads = {}
    dz = y_pred - y_true
    dW2 = (1/m) * np.dot(dz.reshape(1, -1), caches['A1'].T) + (lambda_reg / m) * parameters['W2']
    db2 = (1/m) * np.sum(dz)
    grads['dW2'] = dW2
    grads['db2'] = db2
    dz1 = np.dot(parameters['W2'].T, dz) * (caches['A1'] > 0)
    dW1 = (1/m) * np.dot(dz1.reshape(-1, 1), X.T) + (lambda_reg / m) * parameters['W1']
    db1 = (1/m) * np.sum(dz1)
    grads['dW1'] = dW1
    grads['db1'] = db1

    return grads
```

### 2.3 Dropout  
**Theoretical Explanation**  
- Dropout randomly "drops out" a fraction of neurons during each forward pass, reducing the network's dependency on specific neurons and improving its ability to generalize.

#### Problem 6: Forward Propagation with Dropout <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** During forward propagation, randomly drop units according to the dropout rate to prevent overfitting.  
**Exercise:** Implement `forward_propagation_with_dropout()`.

```python
# In this exercise, you need to apply dropout during forward propagation.
# Hint: Create a mask D with random values, where a fraction 'keep_prob' of the neurons is kept.
# Don't forget to scale the activations after applying the mask to avoid affecting the output values.
# You can use np.random.rand() to generate the dropout mask and multiply element-wise with the activations.

def forward_propagation_with_dropout(X, parameters, keep_prob):
    W1 = parameters['W1']
    b1 = parameters['b1']
    Z1 = np.dot(W1, X) + b1
    A1 = np.maximum(0, Z1)
    D = np.random.rand(*A1.shape) < keep_prob
    A1_dropout = A1 * D / keep_prob

    return A1_dropout, D
```

#### Problem 7: Backward Propagation with Dropout <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Adjust the backward propagation algorithm to handle dropout. You'll compute the gradients only for the units that were not dropped.  
**Exercise:** Implement `backward_propagation_with_dropout()`.

```python
# During backpropagation, apply the same mask D that was used during forward propagation to the gradients.
# Hint: Multiply the gradient of the activation by the same mask D to only propagate through neurons that were kept.

def backward_propagation_with_dropout(dA, D, keep_prob, parameters):
    W1 = parameters['W1']
    m = dA.shape[1]
    dA = dA * D / keep_prob
    dZ1 = dA * (W1 > 0)
    dW = np.dot(dZ1, dA.T) / m
    db = np.sum(dZ1, axis=1, keepdims=True) / m
    dA_prev = np.dot(W1.T, dZ1)

    return dA_prev, dW, db
```

## Section 3: Gradient Checking <span style='color:blue'><strong>(15 Marks)</strong></span>

### 3.1 Gradient Checking Overview  
**Theoretical Explanation**  
- Gradient checking is a technique to verify the correctness of your backward propagation implementation. By comparing the analytical gradient (from backpropagation) with the numerical gradient (using a small perturbation), you can ensure your backpropagation is working correctly.

#### Problem 8: Forward Propagation <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Compute the forward pass for gradient checking. This step is used to compare the numerical gradient later.  
**Exercise:** Implement `forward_propagation()`.


```python
def relu(Z):
    return np.maximum(0, Z)

def sigmoid(Z):
    return 1 / (1 + np.exp(-Z))

def forward_propagation(X, parameters):
    W1 = parameters['W1']
    b1 = parameters['b1']
    W2 = parameters['W2']
    b2 = parameters['b2']
    Z1 = np.dot(W1, X) + b1
    A1 = relu(Z1)
    Z2 = np.dot(W2, A1) + b2
    A2 = sigmoid(Z2)
    cache = {
        'Z1': Z1,
        'A1': A1,
        'Z2': Z2,
        'A2': A2,
        'X': X
    }

    return A2, cache
```

#### Problem 9: Backward Propagation <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Implement the backward propagation algorithm to compute the analytical gradients.  
**Exercise:** Implement `backward_propagation()`.

```python
def compute_cost(A2, Y):
    m = Y.shape[1]
    cost = -np.sum(Y * np.log(A2) + (1 - Y) * np.log(1 - A2)) / m
    return cost

def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))

def backward_propagation(X, Y, parameters, cache):
    m = X.shape[1]
    A1 = cache['A1']
    A2 = cache['A2']
    Z1 = cache['Z1']
    Z2 = cache['Z2']
    dZ2 = A2 - Y
    dW2 = np.dot(dZ2, A1.T) / m
    db2 = np.sum(dZ2, axis=1, keepdims=True) / m
    dZ1 = np.dot(parameters['W2'].T, dZ2) * sigmoid_derivative(Z1)
    dW1 = np.dot(dZ1, X.T) / m
    db1 = np.sum(dZ1, axis=1, keepdims=True) / m
    grads = {
        'dW1': dW1,
        'db1': db1,
        'dW2': dW2,
        'db2': db2
    }

    return grads
```

#### Problem 10: Gradient Check <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Implement gradient checking by computing the numerical gradient and comparing it with the analytical gradient from backpropagation.  
**Exercise:** Implement `gradient_check()`.

```python
# To implement gradient checking, compute the numerical gradient using a small perturbation epsilon.
# Hint: Iterate over each parameter and use (f(theta+epsilon) - f(theta-epsilon)) / (2*epsilon) to compute numerical gradients.
# Compare the numerical gradient with the backpropagation gradient to verify correctness.

def gradient_check(parameters, X, Y, epsilon=1e-7):
    A2, cache = forward_propagation(X, parameters)
    grads = backward_propagation(X, Y, parameters, cache)
    numerical_grads = {}

    for key in parameters:
        original_value = parameters[key]
        param_size = original_value.shape
        numerical_grad = np.zeros(param_size)

        for i in range(param_size[0]):
            for j in range(param_size[1]):
                perturbation = np.zeros(param_size)
                perturbation[i, j] = epsilon
                parameters[key] = original_value + perturbation
                A2_plus = forward_propagation(X, parameters)[0]
                cost_plus = compute_cost(A2_plus, Y)
                parameters[key] = original_value - perturbation
                A2_minus = forward_propagation(X, parameters)[0]
                cost_minus = compute_cost(A2_minus, Y)
                numerical_grad[i, j] = (cost_plus - cost_minus) / (2 * epsilon)
        parameters[key] = original_value
        numerical_grads[key] = numerical_grad

    for key in grads:
        grad_diff = np.linalg.norm(numerical_grads[key] - grads[key]) / (np.linalg.norm(numerical_grads[key]) + np.linalg.norm(grads[key]))
        print(f"Gradient check for {key}: {grad_diff}")

    return grad_diff
```

## Section 4: Optimization Algorithms <span style='color:blue'><strong>(25 Marks)</strong></span>

### 4.1 Optimization Techniques  
**Theoretical Explanation**  
- Gradient descent, momentum, and Adam optimization are techniques used to update the parameters of a neural network. Each method has its benefits and drawbacks in terms of convergence speed and stability.

#### Problem 11: Update Parameters with Gradient Descent <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Implement standard gradient descent, where parameters are updated based on the gradient of the cost function.  
**Exercise:** Implement `update_parameters_with_gd()`.

```python
def update_parameters_with_gd(parameters, grads, learning_rate):
    parameters['W1'] -= learning_rate * grads['dW1']
    parameters['b1'] -= learning_rate * grads['db1']
    parameters['W2'] -= learning_rate * grads['dW2']
    parameters['b2'] -= learning_rate * grads['db2']

    return parameters
```

#### Problem 12: Mini-Batch Gradient Descent <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Split your dataset into mini-batches to increase training efficiency and handle large datasets.  
**Exercise:** Implement `random_mini_batches()`.

```python
# Shuffle the training examples and then divide them into mini-batches of the specified size.
# Hint: You can use np.random.permutation() to shuffle the dataset and split it into mini-batches.

def random_mini_batches(X, Y, mini_batch_size=64, seed=0):
    np.random.seed(seed)
    m = X.shape[0]
    permutation = np.random.permutation(m)
    shuffled_X = X[permutation]
    shuffled_Y = Y[permutation]
    mini_batches = []
    for i in range(0, m, mini_batch_size):
        mini_batch_X = shuffled_X[i:i + mini_batch_size]
        mini_batch_Y = shuffled_Y[i:i + mini_batch_size]
        mini_batches.append((mini_batch_X, mini_batch_Y))

    return mini_batches
```

```python
# Initialize velocity for each parameter as zeros with the same shape as the corresponding parameter.
# Hint: Use np.zeros_like() for initializing VdW and Vdb to store velocity for each parameter.

def initialize_velocity(parameters):
    V = {}
    for key in parameters:
        V[key] = np.zeros_like(parameters[key])

    return V
```

#### Problem 14: Update Parameters with Momentum <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Update parameters using momentum, which incorporates past gradients to smooth the optimization process.  
**Exercise:** Implement `update_parameters_with_momentum()`.

```python
def update_parameters_with_momentum(parameters, grads, V, beta, learning_rate):
    for key in parameters:
        V[key] = beta * V[key] + (1 - beta) * grads[key]
        parameters[key] -= learning_rate * V[key]
    return parameters, V
```

#### Problem 15: Adam Optimization <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Adam is an adaptive optimization algorithm that combines momentum and RMSprop to provide an efficient update rule for parameters.  
**Exercise:** Implement `initialize_adam()` and `update_parameters_with_adam()`.

```python
# Initialize the first and second moment variables (v, s) for Adam optimization.
# Hint: Use np.zeros_like() to initialize both v and s for each parameter.

def update_parameters_with_adam(parameters, grads, v, s, t, learning_rate, beta1, beta2, epsilon):
    for key in parameters:
        v[key] = beta1 * v[key] + (1 - beta1) * grads[key]
        s[key] = beta2 * s[key] + (1 - beta2) * (grads[key] ** 2)
        v_corrected = v[key] / (1 - beta1 ** t)
        s_corrected = s[key] / (1 - beta2 ** t)
        parameters[key] -= learning_rate * v_corrected / (np.sqrt(s_corrected) + epsilon)

    return parameters, v, s
```

## Section 5: Learning Rate Decay and Scheduling <span style='color:blue'><strong>(10 Marks)</strong></span>

### 5.1 Learning Rate Scheduling  
**Theoretical Explanation**  
- The learning rate controls how quickly or slowly your model converges. Learning rate decay reduces the learning rate as training progresses, helping the model converge more stably.

#### Problem 16: Learning Rate Decay per Iteration <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Apply learning rate decay after each iteration to gradually reduce the learning rate during training.  
**Exercise:** Implement `update_lr()`.

```python
# Implement learning rate decay by multiplying the initial learning rate with a decay factor.
# Hint: You can use decay_rate / (1 + decay * epoch) to reduce the learning rate over time.

def update_lr(initial_lr, decay_rate, decay, epoch):
    updated_lr = initial_lr * (decay_rate / (1 + decay * epoch))
    return updated_lr
```

#### Problem 17: Fixed Interval Scheduling <span style='color:blue'><strong>(5 Marks)</strong></span>  
**Guide:** Use a fixed interval scheduling method to reduce the learning rate at specific epochs or intervals.  
**Exercise:** Implement `schedule_lr_decay()`.

```python
# Implement fixed interval scheduling for learning rate decay.
# Hint: Update the learning rate every 'step' iterations by multiplying it with a decay factor.

def schedule_lr_decay(initial_lr, decay_factor, step, epoch):
    intervals_passed = epoch // step
    updated_lr = initial_lr * (decay_factor ** intervals_passed)

    return updated_lr
```

```python
# Use TensorFlow's GradientTape to compute the gradient of the loss function with respect to the parameters.
# Hint: Wrap the forward propagation and loss calculation in a `with tf.GradientTape()` block.

import tensorflow as tf

def linear_function(X, W, b):
    with tf.GradientTape() as tape:
        Y_pred = tf.matmul(X, W) + b
        loss = tf.reduce_mean(tf.square(Y_pred - 1))
    gradients = tape.gradient(loss, [W, b])

    return loss, gradients
```

```python
def initialize_parameters(layer_dims):
    parameters = {}
    num_layers = len(layer_dims)
    for l in range(1, num_layers):
        parameters[f'W{l}'] = tf.random.normal(shape=(layer_dims[l], layer_dims[l-1]), stddev=0.01)
        parameters[f'b{l}'] = tf.zeros(shape=(layer_dims[l], 1))

    return parameters
```
